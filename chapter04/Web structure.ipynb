## Auto Data Collection

Web Development: web page is consisted of HTML, CSS, Javascript

순서
1. Install Live Server (VS code Extension)
2. HTML Structure
3. retrieve text image 
4. save data frame as image file

HTML Structure
[source](index.html)

| Tag | meaning |
| --- | --- |
| `<html>` | HTML 문서 전체 |
| `<head>` | HTML 문서 정보 (화면 표현 x) |
| `<body>` | HTML 문서 페이지 화면에 보이는 내용 |
| `<title>` | 웹브라우저 탭 제목 |
| `<h1> ~ <h6>` | 제목 1 ~ 6까지 |
| `<p>` | 문단 |
| `<a>` | 링크/페이지 이동 |
| `<img>` | 이미지 |
| `<ul>`, `<li>` | 목록, 번호목록 |
| `<table>` | 표 |
| `<input>` | 텍스트 입력창, 콤보박스, .. |
| `<button>` | 버튼 |
| `<div>`, `<span>`, `<section>`, `<content>`, ... | 영역구분 태그 |


```html
<div class="product" id="item-1">
    <span class="name">키보드</span>
    <span class="price">20000</span>
</div>
```
| 속성 | 의미 |
|---|---|
| `id` | 페이지 내 중복되지 않은 값 |
| `class` | CSS 디자인을 위해서 사용. 여러 요소에 공통으로 붙이는 이름. 중복됨 |
| `href` | 링크 주소 |
| `src` | 이미지나 소스 위치 |
| `name` | 입력 요소 이름. 중복됨 |


개발자 도구
- right click > 검사
- element tab > 필요요소 찾기
- check HTML structure
- find location for data scrapping
- check tag name, id, class

선택자
- tag name: h1, span, ul.. 
- .class name: 태그 내 class 속성값 선택
- #id name: 태그 내 id 속성값 선택
