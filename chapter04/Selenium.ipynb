## Automated Web Page Data Collection
- A Python library/package for automatically controlling a web browser.

### What Selenium Can Do
- Launch a web browser
- Open and navigate to web pages
- Enter information into input fields
  - e.g. username, password, search terms
- Click buttons
- Move between pages
- Extract text displayed on a webpage
- Take screenshots of:
  - the entire page
  - a specific image or section
- Automate repetitive browser tasks

### Web Scraping Tools

- **Static HTML pages**
  - `BeautifulSoup4`
  - `requests`

- **Dynamic web pages**
  - `Selenium`
  - Useful when content is loaded or changed through JavaScript

```python
# CSS 선택자로 요소찾는 함수용 패키지 클래스 로드
from selenium.webdriver.common.by import By
```

```python
# retrieve One data from page
title = driver.find_element(By.ID, 'search-btn')

title.text
#element is used for single element like id, elements is for multiple like class
```

```python
# retrieve Multiple elements 
products = driver.find_elements(By.CLASS_NAME, 'product')

# "products 안에 있는 요소를 하나씩 꺼내서 product라고 부르고, 그 요소의 text를 출력해라."
for product in products:
    print(product.text)
```

##### Implicit Wait: 웹 크롤링이나 자동화 테스트(Selenium 등)에서 웹 드라이버가 특정 조건이 충족될 때까지 실행을 멈추고 기다리는 방식
 - link click to go another page
 - `time.sleep(2)` 2seconds wait

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python

wait = WebDriverWait(driver, 10) #UNTIL 10 seconds wait

title = wait.until(
    EC.presence_of_element_located((By.ID, 'title'))
)

title.text
```

Web scrapping:
- web site open with web driver
- search necessary element(s)
- search input tag
- search button element
- execute with input and button click
- retrieve data
- explicit wait until the screen refresh

```python
products = driver.find_elements(By.CLASS_NAME, 'product')

result = []

# can use either CSS.SELECTOR or CLASS_NAME but make sure the format difference
for product in products:
    name = product.find_element(By.CSS_SELECTOR, '.name').text
    price = product.find_element(By.CLASS_NAME, 'price').text

    result.append({'name': name, 'price': price})
```

```python
result
```

```python
df = pd.DataFrame(result)
df
```

```python
df.to_csv('../data/scrap_products.csv', encoding='utf-8', index=False)
```

Practice: Melon Site Top 100 Scrapping

```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python
driver = webdriver.Chrome()
wait = WebDriverWait(driver, 10)
driver.get('https://www.melon.com/chart/index.htm')

Company_name = wait.until(
    EC.presence_of_element_located((By.ID, 'footer'))
)
```

```python
songs = driver.find_elements(By.CSS_SELECTOR, '.lst50, .lst100')
# songs = driver.find_elements(By.CSS_SELECTOR, 'tr.lst50, tr.lst100') 도 가능


# len(songs)
for song in songs:
    # 노래 순위
    rank = song.find_element(By.CSS_SELECTOR, '.rank').text

    # 노래제목
    title = song.find_element(By.CSS_SELECTOR, '.rank01 > span > a').text

    # 가수명
    singer = song.find_element(By.CSS_SELECTOR, '.rank02 a').text

    # 앨범명
    album = song.find_element(By.CSS_SELECTOR, '.rank03 a').text

    # 좋아요 개수
    like = song.find_element(By.CLASS_NAME, 'cnt').text

    # 단계상승하락
    rank_ud = song.find_element(By.CSS_SELECTOR, '.rank_wrap').get_attribute('title')

    # 이미지 썸네일 주소
    # img_url = song.find_element(By.CSS_SELECTOR, '.image_typeAll img').get_attribute('src')

    print(rank, rank_ud, title, singer, album, like)
```

```python
songs = driver.find_elements(By.CSS_SELECTOR, '.lst50, .lst100')

datas = []

# len(songs)
for song in songs:
    # 노래 순위
    rank = song.find_element(By.CSS_SELECTOR, '.rank').text

    # 랭크 순위
    rank_ud = song.find_element(By.CSS_SELECTOR, '.rank_wrap').get_attribute('title')

    # 노래제목
    title = song.find_element(By.CSS_SELECTOR, '.rank01 > span > a').text

    # 가수명
    singer = song.find_element(By.CSS_SELECTOR, '.rank02 a').text

    # 앨범명
    album = song.find_element(By.CSS_SELECTOR, '.rank03 a').text

    # 좋아요 개수
    like = song.find_element(By.CLASS_NAME, 'cnt').text

    print(rank, title, singer, album, like)

    datas.append({
        '순위': rank,
        '랭크순위': rank_ud,
        '노래제목': title,
        '가수': singer,
        '앨범': album,
        '좋아요': like
    })
```

```python
driver.quit()
```

```python
df_melon100 = pd.DataFrame(datas)
df_melon100
```

```python
df_melon100.to_csv('../data/멜론Top100_20161001.csv', encoding= 'utf-8')
```
