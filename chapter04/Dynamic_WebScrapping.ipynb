Dynamic Web Scraping

- Dynamic websites require more steps and resources to scrape.
- Typical process:
  **Open → Search → Scrape → Next Page → Repeat**
Scraping Difficulty

- **Danawa — Medium**
  - JavaScript-rendered pages
  - May restrict User-Agent access

- **Naver — High**
  - Login may be required
  - Changing page structure
  - Repeated requests may be blocked

- **Google — High**
  - Dynamic pages
  - Repeated scraping may trigger CAPTCHA

- **Coupang — Very High**
  - Strong anti-bot protection
  - May return `403 Forbidden`
  - Can block Selenium access

> **Key Point:** Selenium handles dynamic pages, but it cannot guarantee access when anti-bot systems are used.
