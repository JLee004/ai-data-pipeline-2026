Dynamic Web Scraping

- Dynamic websites require more steps and resources to scrape.
- Typical process:
  **Open → Search → Scrape → Next Page → Repeat**
Scraping Difficulty

- **Danawa — Medium**
  - JavaScript-rendered pages
  - May restrict User-Agent access

- **Naver — High**
  - Login may be required
  - Changing page structure
  - Repeated requests may be blocked

- **Google — High**
  - Dynamic pages
  - Repeated scraping may trigger CAPTCHA

- **Coupang — Very High**
  - Strong anti-bot protection
  - May return `403 Forbidden`
  - Can block Selenium access

> **Key Point:** Selenium handles dynamic pages, but it cannot guarantee access when anti-bot systems are used.

Practice site: https://quotes.toscrape.com/

```python
import pandas as pd

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python
driver = webdriver.Chrome()
wait = WebDriverWait(driver, 10)

driver.get('https://quotes.toscrape.com/')

footer = wait.until(
    EC.presence_of_element_located((By.CLASS_NAME, 'footer'))
)
```

wait.until() : 다음 페이지 화면이 전환됐을때 화면 표시되기전 스크래핑 시작을 하면 오류 발생.
time.sleep(10) 보다 효율적

```python
#login

login_link = wait.until(
    EC.element_to_be_clickable((By.LINK_TEXT, 'Login'))
)

login_link.click()
```

##### 여러 페이지 수집

```python
rows = []

driver.get('https://quotes.toscrape.com/')

for page in range(1, 6): 
    wait.until(
        EC.presence_of_all_elements_located((By.CLASS_NAME, 'footer'))
    )

    quotes = driver.find_elements(By.CSS_SELECTOR, '.quote')

    for quote in quotes: # 10 per one page
        text = quote.find_element(By.CSS_SELECTOR, '.text').text
        author = quote.find_element(By.CSS_SELECTOR, '.author').text
        link = quote.find_elements(By.CSS_SELECTOR, 'span > a')[1].get_attribute('href')
        tags = quote.find_elements(By.CSS_SELECTOR, '.tags > a')
        tag_text = ', '.join(tag.text for tag in tags)

        rows.append({
        'text': text,
        'author': author,
        'link': link,
        'tags': tag_text
        })

    next_button = driver.find_element(By.CSS_SELECTOR, 'li.next a')
    if not next_button:
        break
    next_button.click()
    
```

```python
len(rows)
```

```python
df_quotes = pd.DataFrame(rows)
df_quotes
```
