Conditional Retrieval (SQL Where)

````python
df['score'] >=90
````

````python
df[df['score']>=90]
````

- Calculation/operation is similar to NumPy
- In Python use: and, or, not 
- In NumPy and Panda, use: &, |, ~ 

````python
(df['score']>90) & (df['age']>=21)
````

````python
df[(df['score']>90) & (df['age']>=21)]
````

````python
df[(df['score']>90) | (df['age']>=21)]
````

````python
# Q1
df_products
````

````python
# Q2
df_products[['product' , 'price']]
````

````python
# Q3
df_products['price'] >=50000
df_products [df_products['price'] >=50000]
````

````python
# Q4
df_products['stock'] >= 10
df_products [df_products['stock'] >= 10]
````

````python
# Q5
df_products['total_value'] = df_products['price'] * df_products['stock']

df_products
````
