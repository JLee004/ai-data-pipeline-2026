````python
import numpy as np
````

````python
prices = [10000, 15000, 30000, 25000]
discount_prices = []

for price in prices: 
    discount_prices.append(price * 0.9)

discount_prices
````

````python
np_prices = np.array(prices)
np_discount_prices = np_prices * 0.9

np_discount_prices
np_discount_prices.tolist()
````
